```python
# Implement functions here
"""Author: Arthur Chong"""
"""Hexagonal Ice Cube in X-Y coordinate"""
class Position:
    """Type in the wor number and the column number"""
    def __init__(self, row = 0, column = 0):
        self.row = row
        self.column = column
    
        
    def _position(self):
        # The row of the hexagonal ice cube.
        if self.row == 1:
           return self._if_First_Row()
            
        elif self.row == 2:
            return self._if_Second_Row()
            
        elif self.row == 3:
            return self._if_Third_Row()
        
        elif self.row == 4:
            return self._if_Fourth_Row()
            
        elif self.row == 5:
            return self._if_Fifth_Row()
            
        else:
            print("Invalid Input, Please try again")
            
    def _increment_of_X(self):
        x_increment = 0
        
        for x in range (1, self.column):
            x_increment += 25.0 #mm
        
        return x_increment
        
        
    def _if_First_Row(self):
        # Calculate the x coordinate and y_coordinate in mm for the first row.
        
        initial_x = 28.0 # mm
        
        x_increment = self._increment_of_X() # To increase x based on the number of column
        x_coordinate = x_increment + initial_x
        y_coordinate = 17.0 #mm (need to check)
        
        print(f"\nX_Cordinate {x_coordinate}, Y_Cordinate {y_coordinate}")
        
        return x_coordinate, y_coordinate
    
    def _if_Second_Row(self):
        # Calculate the x coordinate and y_coordinate in mm for the second row.

        initial_x = 16 # need to change
        
        x_increment = self._increment_of_X() # To increase x based on the number of column
        x_coordinate = x_increment + initial_x
        y_coordinate = 40.0 #mm 
        
        print(f"\nX_Cordinate {x_coordinate} mm, Y_Cordinate {y_coordinate} mm")
        
        return x_coordinate, y_coordinate
    
    def _if_Third_Row(self):
        # Calculate the x coordinate and y_coordinate in mm for the third row.

        initial_x = 28.0 #mm
        
        x_increment = self._increment_of_X() # To increase x based on the number of column
        x_coordinate = x_increment + initial_x
        y_coordinate = 60.0 #mm (need to check)
        
        print(f"\nX_Cordinate {x_coordinate} mm, Y_Cordinate {y_coordinate} mm")
        
        return x_coordinate, y_coordinate
    
    def _if_Fourth_Row(self):
        # Calculate the x coordinate and y_coordinate in mm for the fourth row.

        initial_x = 16 # need to change
        
        x_increment = self._increment_of_X() # To increase x based on the number of column
        x_coordinate = x_increment + initial_x
        y_coordinate = 81.0 #mm 
        
        print(f"\nX_Cordinate {x_coordinate} mm, Y_Cordinate {y_coordinate} mm")
        
        return x_coordinate, y_coordinate
    
    
    def _if_Fifth_Row(self):
        # Calculate the x coordinate and y_coordinate in mm for the fifth row.

        initial_x = 28.0 # need to change
        
        x_increment = self._increment_of_X() # To increase x based on the number of column
        x_coordinate = x_increment + initial_x
        y_coordinate = 103.0 #mm (need to check)
        
        print(f"\nX_Cordinate {x_coordinate} mm, Y_Cordinate {y_coordinate} mm")
        
        return x_coordinate, y_coordinate

```
