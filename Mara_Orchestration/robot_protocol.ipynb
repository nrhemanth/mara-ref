```python
import sys
from opentrons import protocol_api, execute, simulate
from opentrons.types import Location, Point
import pandas as pd
import numpy as np
import urllib.request
```

```python
df = pd.read_csv("http://192.168.0.100:8000/conditions.csv")
df
```

```python
#protocol = simulate.get_protocol_api('2.10')
protocol = execute.get_protocol_api('2.19')
## pipette and labware
tiprack = protocol.load_labware('opentrons_96_tiprack_300ul', '4')
pip = protocol.load_instrument('p300_single_gen2', 'right', tip_racks=[tiprack])
stock = protocol.load_labware('corning_6_wellplate_16.8ml_flat',location='8')
plate = protocol.load_labware('opentrons_96_wellplate_200ul_pcr_full_skirt','5')
```

```python
# Run the protocol 
protocol.home()
# Dispense the new volumes
for i in range(len(df)):
    well = df["well"][i]
    volume = df["vol"][i]
    print(well,volume)
    
    pip.pick_up_tip()
    pip.aspirate(volume, stock["A1"])  # Adjust source if needed
    pip.dispense(volume, plate[well])
    pip.return_tip()
```
