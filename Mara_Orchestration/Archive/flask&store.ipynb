```python
from flask import Flask, jsonify
import requests
import threading
import time
app = Flask(__name__)
sent_data_store = []  # Store sent data
def send_data():
    data = {"experiment": "Flask Test", "value": 111}
    sent_data_store.append(data)  # Store what was sent
    try:
        response = requests.post("http://192.168.0.102:48888", json=data)  # Send data to Notebook 1
        print("Sent Data:", data)
        print("Server Response:", response.json())  # Print response from receiver
    except requests.exceptions.ConnectionError:
        print("Failed to send data. Is Notebook 1 running?")
@app.route("/send", methods=["GET"])
def trigger_send():
    send_data()
    return jsonify({"message": "Data sent", "stored_data": sent_data_store})
if __name__ == "__main__":
    print("Flask server running. Visit http://127.0.0.1:5000/send to trigger sending.")
    app.run(port=5000)
```
