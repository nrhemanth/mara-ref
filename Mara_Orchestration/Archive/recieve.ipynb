```python
from http.server import BaseHTTPRequestHandler, HTTPServer
import json
class RequestHandler(BaseHTTPRequestHandler):
    def do_POST(self):
        content_length = int(self.headers['Content-Length'])  # Get content length
        post_data = self.rfile.read(content_length)  # Read data
        data = json.loads(post_data.decode('utf-8'))  # Decode JSON
        print("Received Data:", data)  # Print received data
        # Send response
        self.send_response(200)
        self.send_header('Content-type', 'application/json')
        self.end_headers()
        self.wfile.write(json.dumps({"message": "Data received"}).encode('utf-8'))
# Start the HTTP server
server_address = ('', 6000)  # Listen on port 6000
httpd = HTTPServer(server_address, RequestHandler)
print("Listening for incoming data on port 6000...")
httpd.serve_forever()  # Keep running
```
