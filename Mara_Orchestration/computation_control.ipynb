```python
import requests
import random

robot_ip = "192.168.1.100"  # Replace with your Opentrons robot IP
url = f"http://{robot_ip}:5000/run_protocol"

# Simulate Bayesian Optimization results
optimized_well = random.choice(["A1", "B2", "C3"])
optimized_volume = random.randint(20, 100)

# Send HTTP request
data = {"well": optimized_well, "volume": optimized_volume}
response = requests.post(url, json=data)

print(f"Response from robot: {response.json()}")
```
