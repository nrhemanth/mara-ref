```python
import asyncio
import websockets
import json

async def send_variable(var_name, var_value):
    uri = f"ws://127.0.0.1:8765"
    async with websockets.connect(uri) as ws:
        data = {
            "var_name": var_name,
            "var_value": var_value
        }
        await ws.send(json.dumps(data))
        
        response = await ws.recv()
        print("response from server:", response)
        
await send_variable("motor_speed", 250)
await send_variable("solution_volume", [100,200,300])
```
