```python
import asyncio
import websockets
import json
import random

async def send_data():
    uri = "ws://localhost:8765"
    async with websockets.connect(uri) as websocket:
        optimize_well = random.choice(["A1","B2","C3"])
        optimize_volume = random.randint(20,100)
        
        data = {"well": optimize_well,"volume":optimize_volume}
        await websocket.send(json.dumps(data))
        print(f"Sent Optimzied Parameters")
        
```
