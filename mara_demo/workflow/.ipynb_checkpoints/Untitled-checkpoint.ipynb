```python



# ==================================================
# CLIENT CODE - Run this on localhost (localhost:8888)
# ==================================================

import asyncio
import websockets
import json

class OpentronsBridge:
    def __init__(self, robot_ip="196.254.194.251", port=9000):
        self.robot_ip = robot_ip
        self.port = port
        self.websocket = None
    
    async def connect(self):
        """Connect to the Opentrons server"""
        uri = f"ws://{self.robot_ip}:{self.port}"
        try:
            self.websocket = await websockets.connect(uri)
            print(f"Connected to Opentrons at {uri}")
            return True
        except Exception as e:
            print(f"Connection failed: {e}")
            return False
    
    async def disconnect(self):
        """Close the connection to Opentrons"""
        if self.websocket:
            await self.websocket.close()
            print("Disconnected from Opentrons")
    
    async def send_message(self, message):
        """Send a message to Opentrons and get the response"""
        if not self.websocket:
            print("Not connected to Opentrons")
            return None
        
        try:
            # Send the message
            await self.websocket.send(message)
            print(f"Sent to Opentrons: {message}")
            
            # Wait for a response
            response = await self.websocket.recv()
            print(f"Received from Opentrons: {response}")
            return response
        except Exception as e:
            print(f"Communication error: {e}")
            return None

# Test function to run in a cell
async def test_opentrons_connection():
    # Connect to Opentrons
    bridge = OpentronsBridge()
    connected = await bridge.connect()
    
    if connected:
        try:
            # Test message 1
            await bridge.send_message("Hello from localhost!")
            
            # Test message 2
            await bridge.send_message("This is a test connection")
            
            # Test JSON
            data = {
                "command": "status",
                "parameters": {"check": "temperature"}
            }
            await bridge.send_message(json.dumps(data))
            
        finally:
            # Close the connection
            await bridge.disconnect()
    else:
        print("Could not establish connection")

# Run the test in a cell on localhost
def run_localhost_client():
    try:
        loop = asyncio.get_event_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)
    
    loop.run_until_complete(test_opentrons_connection())

# Execute in localhost notebook:
# run_localhost_client()


"""
INSTRUCTIONS:

1. On your Opentrons notebook (196.254.194.251:48888):
   - Install websockets if needed: !pip install websockets
   - Run the SERVER CODE section in a cell
   - Execute: server_task = run_opentrons_server()
   - The server will start on port 9000 (you can change this if needed)

2. On your localhost notebook (localhost:8888):
   - Install websockets if needed: !pip install websockets
   - Run the CLIENT CODE section in a cell
   - Execute: run_localhost_client()
   - This will connect to Opentrons and exchange some test messages

3. For actual use:
   - Create your own methods in the OpentronsBridge class for specific functionality
   - You can send commands, data, or code to execute between the two notebooks

Note: If you get connection errors, make sure:
- Both notebooks have websockets installed
- Port 9000 is not being blocked by a firewall
- The IP address 196.254.194.251 is correctly set
"""
```
