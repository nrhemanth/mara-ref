```python
import serial

# Change 'COM3' to your correct serial port (Windows) or '/dev/ttyUSBx' on Linux
SERIAL_PORT = "/dev/ttyACM0"  # Use the port from Device Manager or `/dev/ttyUSB0`
BAUD_RATE = 115200

ser = serial.Serial(SERIAL_PORT, BAUD_RATE, timeout=1)

# Example: Send a command
def send_command(command):
    """Send a command to Opentrons via Serial."""
    ser.write(f"{command}\n".encode('utf-8'))  # Send command
    response = ser.readline().decode('utf-8').strip()  # Read response
    print("Response from Opentrons:", response)

# Test Commands
send_command("PING")  # Opentrons should respond with "PONG"
send_command("DISPENSE:200")  # Opentrons will print "Dispensing 200μL"
```
