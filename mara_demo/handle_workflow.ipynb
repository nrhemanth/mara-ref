```python
import asyncio
import websockets
import json

shared_variables = {}

async def handle_connection(websocket, path):
    async for message in websocket:
        data = json.loads(message)
        
        var_name = data.get("var_name", "")
        var_value = data.get("var_value", None)
        
        shared_variables[var_name] = var_value
        print(f"stored: {var_name} = {var_value}")
        
        response = {
            "status": "ok",
            "message": f"stored {var_name} successfully"
        }
        await websocket.send(json.dumps(response))
        
async def main_server():
    server = await websockets.serve(handle_connection, "0.0.0.0", 8765)
    print("websocket server running on port 9876...")
    await server.wait_closed()
    
await main_server()
```
